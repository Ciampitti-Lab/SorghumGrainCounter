# Segmentation

## Loading Models

```python
from ultralytics import YOLO

model = YOLO("yolo12n-seg.pt")
```

## Training

```python
results = model.train(data      =   "data/yolo_cropped/data.yaml",
                      epochs    =   100,
                      imgsz     =   640,
                      patience  =   10,
                      batch     =   16,
                      device    =   0,
                      max_det   =   1,
                      plots     =   False)
```

## Validation

```python
from ultralytics import YOLO

# Load a model
model = YOLO("models/yolov8n.pt")

# Validate the model
metrics = model.val()  # no arguments needed, dataset and settings remembered
metrics.seg.map  # map50-95(M)
metrics.seg.map50  # map50(M)
metrics.seg.map75  # map75(M)
metrics.seg.maps  # a list contains map50-95(M) of each category
```

## Inference

```python
from ultralytics import YOLO

# Load a model
model = YOLO("models/yolo11n.pt")

# Predict with the model

results = model.predict(source=["cropped_image_0.jpg", "cropped_image_1.jpg"], device=0, max_det=1, conf=0.8, save = True)
```
